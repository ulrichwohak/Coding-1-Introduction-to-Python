# Data munging
We have data about hotels in Vienna and we have one goal for today:
We want to find a hotel room for at most 100EUR with a rating >=4.

We also want to inspect our data (DataFrame):
1. What information do we have and what does each row represent?
2. Are prices for a single night?
3. What does a typical night cost? How much do prices vary?
4. Which rooms meet our budget?
6. Which rooms are cheapest per night?

```python
import pandas as pd
import numpy as np

# Import data
df = pd.read_csv("https://raw.githubusercontent.com/ulrichwohak/Coding-1-Introduction-to-Python/main/lectures/lecture04-pandas-basics/hotel_vienna_quotes.csv")

# Get a glimpse of the first few rows
df.head()
```

```python
# DataFrame also have properties/attributes
df.columns # Notice we are not calling columns with ()
# It is a property, not a method!

# Let's not keep all of them
cols_to_keep = ['hotel_id', 'neighbourhood', 'stars', 'price', 'nnights',
                'ratings', 'rating_count', 'distance', 'ratingta']

# Let's just grab those columns
hotels = df[cols_to_keep].copy()
hotels.head()
```

```python
# Excursion:
# Why one would want to use the .copy method.
lst1 = [1, 2, 3]
lst2 = lst1 #.copy()

lst2[0] = 999
print(lst1)
```

```python
# What other options does a DataFrame give me to investigate my data
print(hotels.shape) # Output is of type tuple to prevent manual changes

# hotels.shape[0] = 1
print("Number of rows:", hotels.shape[0])
print(f"Number of rows: {hotels.shape[0]}")

print(f"Number of cols: {hotels.shape[1]}")
```

```python
# Another useful method
hotels.info()
```

```python
# Let's check how many unique hotel_id values we have
# There's two ways to select a single column
# hotels.hotel_id
unique_hotel_ids = hotels["hotel_id"].nunique()
print(f"We have {unique_hotel_ids} unique hotel ids.")
```

```python
# How many hotel bookings do we have for each star category?
hotels["stars"].value_counts().sort_index(ascending=False) #.plot(kind="bar")
```

```python
# Let's check the distribution of nnights
hotels["nnights"].value_counts()
```

```python
# Visually inspect the cols of interest
print(hotels[["price", "nnights"]].head())

# Create a new column
hotels["price_per_night"] = hotels["price"] / hotels["nnights"]
hotels[["price", "nnights", "price_per_night"]].head()
```

```python
# General descriptive stats with the describe() method
# Let's look at three columns: price_per_night, ratings, distance
hotels[["price_per_night", "ratings", "distance"]].describe()
```

```python
# We appear to have 2 missing observations for the ratings variable
missing_ratings = hotels[["ratings", "ratingta"]].isna().sum()
print(missing_ratings)
# print(f"We have {missing_ratings} observations for which ratings are missing.")
```

```python
# We do wanna know which hotels have a price of less than 100
within_budget = hotels["price_per_night"] < 100

# We can use this as an index
print(f"We have {hotels.shape[0]} rooms in the data.")
affordable_rooms = hotels[within_budget]
print(f"After filtering, we have {affordable_rooms.shape[0]} observations in our data.")
```

```python
# Find a good deal
good_deals = (hotels["price_per_night"] < 100) & (hotels["ratings"] >= 4)
good_deals_data = hotels.loc[good_deals]
print(len(good_deals_data))
```
