# We can store things in variables
Note that they ALWAYS have a "type".

```python
# Assign a value to a variable
a = 3
print(a)
print(type(a))
```

```python
# There are additional built-int types in Python
b = 3.14 # Float
print(type(b))

# Strings
c = "Hello"
d = 'Hello Mister "Quotation"'

print(d)
```

```python
# Basic arithmetic operations with numeric objects
x = 6
y = 2
print(x + y)  # Addition
print(x - y)  # Subtraction
print(x * y)  # Multiplication
print(x // y)  # IntegerDivision
print(x / y)  # Division; returns float by default
```

```python
# In order to change a float to an int, we can use type casting
a = 6
b = 2
res = 6 / 2
res = int(res)  # Type casting to int
print(type(res))
```

```python
# Let's talk about strings
string1 = "Apple"
string2 = "A"

# Operations on strings
print(string1 + string2) # Concatenation; NOT addition

# Multiplication of strings
print(string2 * 50) # Repetition
print(string2 / 2) # Division is not defined for strings
```

```python
# We can also typcast to str
print(str(5))
```

```python
# Boolean
val1 = True
val2 = False

# Logical statements
# and combinations of logical statements
a = 5
b = 10

print(a < 7) # True
print(b < 9) # False
print(a < 7 and b < 9) # False
print(a < 7 or b < 9) # True


```
