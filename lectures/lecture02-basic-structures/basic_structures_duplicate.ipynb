# Collections
Collections are in-built data types that allow us to group objects

```python
lst1 = [1, 'a', 3.14]
print(lst1)

# Print the first element of lst1
lst1[0] # We index with []
```

```python
lst2 = []

lst2.append(1)
lst2.append(3)
lst2.append(2)
lst2.append(1.5)
print(lst2)
lst2.sort()
print(lst2)
```

```python
# How many elements are in my list?
print(len(lst2))
```

```python
# Slicing
lst2[0:3]
```

```python
# We can also nest lists
lst2.append([1, 2, 3])
print(lst2)
```

```python
print(lst2.sort())
```

```python
# Grabbing the last elements of a list
print(lst2[-1]) # = [1, 2, 3]
print(lst2[-1][-1]) # The last element
```

```python
print(lst2)
lst2[0] = [1, 2, 3] # Lists are mutable objects
print(lst2)
```

```python
# Immutable list-like objects: tuple
t = (1, 2, 3)
t[0] = 5
```

```python
# Multiple ways of defining a tuple
tup1 = 5, 6, 7, '8'
print(tup1)
print(type(tup1))
```

```python
# Dictionary: key-value pairs
d = {'a': 1, 
     'b': 2, 
     'c': 3}
print(d)
print(type(d))
```

```python
# How to access elements in a dictionary
d['a']
```

```python
print('a' in d)
```

```python
# Sets
s = {1, 1, 1, 4, 5}
print(s) # Collection of uniqyue (!!) elements
```

```python
# How could we de-duplicate a list?
lst3 = [1, 1, 1, 4, 5]
lst4 = list(set(lst3))
print(lst4)
```

```python
# We also have mathematical set operations
s1 = {1, 2, 3}
s2 = {2, 3, 4}

# Union
print(s1 | s2)
print(s1.union(s2))

# Intersection
print(s1 & s2)
print(s1.intersection(s2))
```
