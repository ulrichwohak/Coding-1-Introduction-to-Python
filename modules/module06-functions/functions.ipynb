## User-defined Functions <a class='anchor' id="UDF"></a>

A function is a block of organized, reusable code that is used to perform a single, related action. Functions provide better modularity for your application and a high degree of code reusing.

You can define functions to provide the required functionality. Here are simple rules to define a function in Python.

* Function blocks begin with the keyword ```def``` followed by the function name and parentheses ```( )```.

* **Parameters** are the names inside the parentheses in a function definition. **Arguments** are the values supplied when calling the function.

* The first statement of a function can be an optional statement - the documentation string of the function or docstring.

* The code block within every function starts with a colon (```:```) and is **indented**.

* The statement ```return``` [expression] returns a value, or a serious of values, a list, a dictionary, .... A return statement with no arguments is the same as return None.

````python
# Define a reusable calculation; defining the function does not run its body yet.
def add_one(number):
    x = number + 1
    return x
````

````python
# Call the function with 20 as its input and display its returned value.
add_one(20)
````

You can return more than one object from a single function. 

````python
# Return two results together; Python packages them as a tuple.
def add_one_and_return_both(number):
    x = number
    y = x + 1
    return x, y
````

````python
# Unpack the two returned values into separate names before printing them.
x, y = add_one_and_return_both(23)
print(x)
print(y)
````

## Positional and keyword arguments <a class="anchor" id="arguments"></a>

The function below has two **parameters**: `number` and `exponent`.
The **arguments** are the values we supply when we call it.

`exponent=2` in the definition sets a **default value**: if we leave out the
exponent when calling the function, it uses `2`. A default value does not make
a parameter keyword-only.

````python
# Supply a default exponent so callers can omit it when they want a square.
def number_to_the_power(number, exponent=2):
    return number ** exponent
````

````python
# One positional argument: number is 5; exponent uses its default value, 2.
print(number_to_the_power(5))
````

````python
# Positional arguments are matched by order: number first, exponent second.
print(number_to_the_power(5, 3))
# Swapping the values changes the calculation: 3 to the power of 5.
print(number_to_the_power(3, 5))
````

````python
# Keyword arguments are matched by parameter name, not their order in the call.
print(number_to_the_power(number=5, exponent=3))
print(number_to_the_power(exponent=3, number=5))
````

````python
# We can mix the two styles: positional arguments first, then keyword arguments.
print(number_to_the_power(5, exponent=3))
# A required parameter can also be supplied by keyword; exponent still defaults to 2.
print(number_to_the_power(number=5))
````

**What to remember**

- For this function, both parameters accept positional or keyword arguments.
- With positional arguments, order matters. With keyword arguments, names decide
  which parameter receives each value.
- In these ordinary calls, put positional arguments before keyword arguments,
  use the exact parameter names, and do not supply the same parameter twice.

These calls are **invalid** (shown as text, not runnable cells):

```python
number_to_the_power(number=5, 3)  # A positional argument follows a keyword argument.
number_to_the_power(5, number=3)  # number receives two values; exponent is not set to 3.
```

See the [Python tutorial on keyword arguments](https://docs.python.org/3.12/tutorial/controlflow.html#keyword-arguments).

## Docstrings <a class="anchor" id="docstrings"></a>

A **docstring** is a string placed as the first statement inside the function.
It documents what the function does; `help(function_name)` displays it.

[PEP 257](https://peps.python.org/pep-0257/) gives Python's general conventions:
use triple double quotes (`"""`), start with a short command-style summary such
as "Convert ...", and separate a longer explanation from that summary with a
blank line. Explain inputs, returned values, and important side effects.

The example below uses [NumPy-style docstrings](https://numpydoc.readthedocs.io/en/latest/format.html)
with `Parameters` and `Returns` sections. Those headings are a documentation
convention, not mandatory Python syntax. For a simple function, a one-line
docstring such as `"""Return the input number plus one."""` is enough.

Here the function returns a list, but also changes the supplied list itself.
The docstring must make that side effect clear.

````python
# Convert items in place; use items rather than masking Python's built-in list name.
def cast_listitems_to_string(items):
    """Convert each item to a string in place and return the same list.

    Parameters
    ----------
    items : list
        List whose elements will be replaced by their string representations.

    Returns
    -------
    list
        The same list, now containing strings. No new list is created.
    """
    for i in range(len(items)):
        items[i] = str(items[i])  # Replace the element in the original list.
    return items
````

`help()` displays the function's signature and its docstring without running
the function. Notice that the documentation describes both its returned value
and its effect on the input list.

````python
# help displays the function's signature and docstring without calling the function.
help(cast_listitems_to_string)
````

````python
# Prepare inputs with mixed types to test the conversion function.
import math 

ls_convertable = [1,2, 'a', math.cos(math.pi / 3)]
````

````python
# Inspect the input values before they are converted.
ls_convertable
````

````python
# The function mutates and returns the same list; both variable names now refer to that
# list.
ls_converted = cast_listitems_to_string(ls_convertable)
````

````python
# Inspect the converted elements; numeric-looking strings are still text.
ls_converted
````

## Lambda Functions <a class="anchor" id="lambda"></a>

A lambda function is a small anonymous function. A lambda function can take any number of arguments, but can only have one expression. It is created using the `lambda` keyword.

````python
# A lambda defines a short function consisting of one expression whose value is
# returned.
square = lambda x: x ** 2
````

````python
# Call the lambda like any other function.
square(2)
````

We use lambda to simplify our code, to create temporary definitions, which are used only once. The same can be achieved with a normal definiton:

````python
# Express the same calculation with def and an explicit return statement.
def square_def(x): 
    return x ** 2
````

````python
# Check that the def version gives the same result for the same input.
square_def(2)
````

### Use case: sort hotels by price

Each tuple contains a hotel name and its price. We want to sort by **price**, not
by name. First, write a regular function that returns the price of one hotel.

````python
hotels = [("Central", 120), ("Budget", 70), ("Riverside", 95)]

def get_price(hotel):
    return hotel[1]  # The second element of one tuple is its price.

print(sorted(hotels, key=get_price))
````

**What does `key=get_price` do?**

- We pass the function itself, without parentheses.
- `sorted()` calls `get_price` once for **each hotel tuple**, not on the full list.
  For Central it returns `120`, for Budget `70`, and for Riverside `95`.
- These prices determine the order. The result is a new list of **complete hotel
  tuples**; the original `hotels` list is unchanged.

Now replace `get_price` with a lambda: `lambda hotel: hotel[1]` means
"take one hotel tuple and return its price."

````python
# Supply the same sorting rule directly, without a separate function definition.
print(sorted(hotels, key=lambda hotel: hotel[1]))
````

Both versions print:

```text
[('Budget', 70), ('Riverside', 95), ('Central', 120)]
```

A lambda is useful here because another function needs a short, single-expression
function as an input. Use `def` when the function needs several steps or a helpful
name. See the [Python sorting guide](https://docs.python.org/3.12/howto/sorting.html#key-functions).

You can combine `lambda` functions with *list comprehension*. 

````python
# Create input numbers from zero through nine.
ls_numbers = list(range(10))
````

````python
# Display the list that later comprehensions will process.
ls_numbers
````

Let's square all the values from the list and add 1 to each element

````python
# Apply the same function to every input and collect the results in a new list.
f = lambda x: x**2 + 1
[f(x) for x in ls_numbers]
````

Let's square and add one to each even number in the list

````python
# Filter for even inputs first, then apply the function to those inputs only.
[f(x) for x in ls_numbers if x%2 == 0 ]
````

Square and add one to each even number in the list but return the odd numbers without transformation

````python
# Keep every input position: transform even values but leave odd values unchanged.
[f(x) if x%2 == 0 else x for x in ls_numbers]
````

You can also handle errors with lambda functions and conditional list comprehension.

````python
# Define a text transformation that replaces commas with decimal points.
replace_comma = lambda x: x.replace(',', '.')
````

````python
# The result is still a string; replacing punctuation does not convert it to a number.
replace_comma('4,5')
````

````python
# Prepare mixed text and numeric inputs to demonstrate a type-dependent transformation.
ls_mixed_data = [1.2, '1,2', 5, 7, '4,5', 7]
````

````python
# Only strings have replace applied; numeric values pass through unchanged.
[replace_comma(x) if isinstance(x, str) else x for x in ls_mixed_data]
````
